Question 2 : Explain the difference between List and Tuple in Python.

List :  A list is an ordered collection of   items, where each item can be of any data type (e.g., integers, strings ).
  
  Lists are written using square brackets [].
  * Key Features:

  * Mutable: Items can be added, removed, or changed.
  * Indexed: Items can be accessed using their position in the list.

  **Accessing Items and adding Items in a List:**
1. Accessing item

2. Adding items : Use **.append()** to add an item at the end of the list.

## **Removing and Sorting Items in a List:**

1. Removing Items: Use remove() to delete a specific item.

2. Sorting Items: Use sort() to arrange items in ascending order. It sorts the item alphabetically in string and in ascending or descending order in integers.


* Tuple :  
A tuple is an immutable sequence of items, meaning the elements cannot be changed after the tuple is created.

  Tuples are written using ().

* Key Features:
  * Ordered: Elements maintain their sequence.
  * Immutable: No addition, removal, or modification of items.
  * Allows Duplicates: Like lists, tuples can have duplicate values.

**Why Use Tuples?**

* Use tuples when you need a fixed collection of items that won’t change.

* Suitable for data that must remain constant, such as coordinates, configuration settings, or dictionary keys.

### **Similar to Lists, Tuples also consider indexing.**


## Key Differences between List and Tuple in Python

| Feature           | List                                         | Tuple                                          |
| :---------------- | :------------------------------------------- | :--------------------------------------------- |
| **Mutability**    | Mutable (can be changed after creation)      | Immutable (cannot be changed after creation)   |
| **Syntax**        | Defined using square brackets `[]`           | Defined using parentheses `()`                 |
| **Performance**   | Slower operations due to overhead of mutability | Faster operations as they are fixed in size |
| **Memory**        | Generally consumes more memory               | Generally consumes less memory                 |
| **Use Case**      | Storing collections of items that may change,   | Storing collections of items that should not |
|                   | e.g., a list of users, a shopping cart.      | change, e.g., coordinates, RGB color values.   |
| **Methods**       | Has methods like `append()`, `extend()`, `insert()`, `remove()`, `pop()`, `sort()`, `reverse()` | Has fewer methods, mainly `count()` and `index()` |


**In essence:**
-   **Lists** are for collections that are expected to change frequently.
-   **Tuples** are for collections that should remain constant throughout the program's execution.

Question 3 : What is a function in Python? Why are functions useful?

## **Functions**

  A function is a block of organized, reusable code that performs a specific task.

* Purpose:

  To enable modularity, reduce redundancy, and improve code clarity.
 ## **Types of Function**

1. Built-in Functions
2. User-defined Functions
3. Lambda (Anonymous) Functions
4. Recursive Functions
5. Higher Order functions

*1. Built-in Functions*

* Already provided by Python, ready to use.

* Examples: print(), len(), type(), sum(), max().

*2. User-defined Functions*  

* Functions you create yourself using the def keyword.

*3. Lambda Functions*

* Small, one-line functions defined with lambda.

* Often used for short operations or as arguments to higher-order functions.

*4. Recursive Functions*

* Functions that call themselves to solve problems step by step.

*5. Higher Order Functions*

* Functions that take other functions as arguments or return them.

* Examples: map(), filter(), reduce().

Question 5 : Write a Python program to check whether a number is even or odd using conditional
statements.

```python
# Take a number as input from the user
num_str = input("Enter an integer: ")

# Convert the input string to an integer
try:
    number = int(num_str)

    # Check if the number is even or odd using the modulo operator
    if number % 2 == 0:
        print(f"The number {number} is even.")
    else:
        print(f"The number {number} is odd.")
except ValueError:
    print("Invalid input. Please enter a valid integer.")
```

Question 8 : Write a Python program to remove duplicate values from a list using a set.

```python
# Original list with duplicate values
original_list = [1, 2, 2, 3, 4, 4, 5, 1]

print(f"Original list: {original_list}")

# Convert the list to a set to remove duplicates
# Sets automatically store only unique elements
unique_elements_set = set(original_list)

# Convert the set back to a list (if a list is desired as output)
list_without_duplicates = list(unique_elements_set)

print(f"List after removing duplicates: {list_without_duplicates}")
```

Question 9 : Write a function that returns the square of a number.

```python
def square(number):
    """This function returns the square of the given number."""
    return number * number

# Demonstrate the function
input_number = 7
result = square(input_number)
print(f"The square of {input_number} is: {result}")
```

Question 10 : Write a Python program to count how many times a number appears in a list.

Example List

[2,3,4,2,5,2]

```python
# Example List
my_list = [2, 3, 4, 2, 5, 2]

# The number to count
number_to_count = 2

# Use the count() method to find occurrences
count = my_list.count(number_to_count)

print(f"The number {number_to_count} appears {count} times in the list.")
```
